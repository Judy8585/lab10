```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2026/2026-02-10/schedule.csv"
df = pd.read_csv(url)

df.head()
```

```python
df.columns
```

```python
venue_counts = (
    df["venue_name"]
    .value_counts()
    .head(10)
)

venue_counts
```

```python
venue_counts.sort_values().plot(kind="barh")

plt.xlabel("Number of Events")
plt.ylabel("Venue")
plt.title("Top 10 Venues Hosting the Most Events")
plt.tight_layout()
plt.show()
```

## Interpretation

Cortina Curling Olympic Stadium hosts the most scheduled events, with 436 events. Livigno Snow Park and Cortina Sliding Centre follow with 297 and 290 events. The top venue hosts substantially more events than the other venues.
